```python
import numpy as np

# 1. Binary Step Function
def binary_step(x):
    return np.where(x >= 0, 1, 0)

print("--- Binary Step Function ---")
print("Formula: f(x) = 1 if x >= 0, else 0")
inputs = np.arange(-10, 11)
outputs = binary_step(inputs)
print("Input (x) | Output f(x)")
print("-----------------------")
for i, o in zip(inputs, outputs):
    print(f"{i:9} | {o:11}")
```

```python
# 2. Sigmoid Function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

print("\n--- Sigmoid Function ---")
print("Formula: f(x) = 1 / (1 + e^(-x))")
inputs = np.arange(-10, 11, dtype=float)
outputs = sigmoid(inputs)
print("Input (x) | Output f(x)")
print("-----------------------")
for i, o in zip(inputs, outputs):
    print(f"{i:9} | {o:11.6f}")
```

```python
# 3. Tanh Function
def tanh(x):
    return np.tanh(x)

print("\n--- Tanh Function ---")
print("Formula: f(x) = (e^x - e^(-x)) / (e^x + e^(-x))")
inputs = np.arange(-10, 11, dtype=float)
outputs = tanh(inputs)
print("Input (x) | Output f(x)")
print("-----------------------")
for i, o in zip(inputs, outputs):
    print(f"{i:9} | {o:11.6f}")
```

```python
# 4. ReLU (Rectified Linear Unit) Function
def relu(x):
    return np.maximum(0, x)

print("\n--- ReLU Function ---")
print("Formula: f(x) = max(0, x)")
inputs = np.arange(-10, 11, dtype=float)
outputs = relu(inputs)
print("Input (x) | Output f(x)")
print("-----------------------")
for i, o in zip(inputs, outputs):
    print(f"{i:9} | {o:11.6f}")
```

```python
# 5. Leaky ReLU Function
def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, x * alpha)

print("\n--- Leaky ReLU Function (alpha=0.01) ---")
print("Formula: f(x) = x if x > 0, else alpha * x")
inputs = np.arange(-10, 11, dtype=float)
outputs = leaky_relu(inputs)
print("Input (x) | Output f(x)")
print("-----------------------")
for i, o in zip(inputs, outputs):
    print(f"{i:9} | {o:11.6f}")
```

```python
# 6. ELU (Exponential Linear Unit) Function
def elu(x, alpha=1.0):
    return np.where(x > 0, x, alpha * (np.exp(x) - 1))

print("\n--- ELU Function (alpha=1.0) ---")
print("Formula: f(x) = x if x > 0, else alpha * (e^x - 1)")
inputs = np.arange(-10, 11, dtype=float)
outputs = elu(inputs)
print("Input (x) | Output f(x)")
print("-----------------------")
for i, o in zip(inputs, outputs):
    print(f"{i:9} | {o:11.6f}")
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np
import time

print("\n--- Task 6: Neural Network Experiment (MNIST) ---")
print("Loading and preprocessing MNIST dataset...")

# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

print(f"Original x_train shape: {x_train.shape}")
print(f"Original y_train shape: {y_train.shape}")

# Normalize pixel values to 0-1 range
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

# Reshape images to vectors (28*28 = 784 pixels)
x_train = x_train.reshape(-1, 28 * 28)
x_test = x_test.reshape(-1, 28 * 28)

# Convert labels to one-hot encoding
y_train = keras.utils.to_categorical(y_train, num_classes=10)
y_test = keras.utils.to_categorical(y_test, num_classes=10)

print(f"x_train (reshaped and normalized) shape: {x_train.shape}")
print(f"y_train (one-hot encoded) shape: {y_train.shape}")
print("Data preprocessing complete.")
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import time

# Function to create a simple neural network model
def create_model(activation_function):
    model = keras.Sequential([
        layers.Input(shape=(784,)),
        layers.Dense(128, activation=activation_function),
        layers.Dense(64, activation=activation_function),
        layers.Dense(10, activation='softmax') # Output layer for 10 classes
    ])

    model.compile(
        optimizer='adam',
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )
    return model

# Common training parameters
epochs = 10
batch_size = 32

# Dictionary to store results for comparison
results = {}
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import time

# Train with Sigmoid activation
print(f"\n--- Training with Sigmoid activation ---")
sigmoid_model = create_model('sigmoid')
sigmoid_model.summary()

start_time = time.time()
history_sigmoid = sigmoid_model.fit(
    x_train, y_train,
    epochs=epochs,
    batch_size=batch_size,
    validation_data=(x_test, y_test),
    verbose=0 # Suppress verbose output for cleaner display
)
end_time = time.time()
training_time_sigmoid = end_time - start_time

# Evaluate and store results
loss_sigmoid, accuracy_sigmoid = sigmoid_model.evaluate(x_test, y_test, verbose=0)
val_accuracy_sigmoid = history_sigmoid.history['val_accuracy'][-1]

results['sigmoid'] = {
    'accuracy': accuracy_sigmoid,
    'loss': loss_sigmoid,
    'val_accuracy': val_accuracy_sigmoid,
    'training_time': training_time_sigmoid
}

print(f"Sigmoid - Test Accuracy: {accuracy_sigmoid:.4f}, Test Loss: {loss_sigmoid:.4f}, Val Accuracy: {val_accuracy_sigmoid:.4f}, Training Time: {training_time_sigmoid:.2f}s")
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Define a generic CNN model builder function
def create_cnn_model(input_shape, num_classes, activation_fn):
    model = keras.Sequential([
        # Convolutional Layer 1
        layers.Conv2D(32, (3, 3), activation=activation_fn, input_shape=input_shape),
        layers.BatchNormalization(),
        layers.MaxPooling2D((2, 2)),

        # Convolutional Layer 2
        layers.Conv2D(64, (3, 3), activation=activation_fn),
        layers.BatchNormalization(),
        layers.MaxPooling2D((2, 2)),

        # Flatten the output for the dense layers
        layers.Flatten(),

        # Dense Layer 1
        layers.Dense(128, activation=activation_fn),
        layers.BatchNormalization(),
        layers.Dropout(0.5), # Added dropout for regularization

        # Output Layer
        layers.Dense(num_classes, activation='softmax') # Softmax for multi-class classification
    ])

    model.compile(
        optimizer='adam',
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )
    return model

# Display a summary of the model for ReLU activation (as an example)
print("Example CNN model summary with ReLU activation:")
mnist_input_shape = mnist_x_train.shape[1:]
mnist_num_classes = mnist_y_train.shape[1]

example_model = create_cnn_model(mnist_input_shape, mnist_num_classes, 'relu')
example_model.summary()

print("CNN model architecture defined.")
```

```python
import numpy as np
import matplotlib.pyplot as plt

print("\n--- Task 8: Build Your Own Activation Function ---")

# Custom Activation Function: Smooth Linear Softplus (SLS)
# This function aims to combine the linearity of ReLU for positive inputs
# with a smooth, non-zero gradient for negative inputs, inspired by Softplus,
# while ensuring differentiability at x=0.

# Formula:
# f(x) = x                  if x >= 0
# f(x) = 2 * log(1 + e^x)   if x < 0
# The factor of 2 ensures differentiability at x=0, as derivative of x is 1,
# and derivative of log(1+e^x) is sigmoid(x). At x=0, sigmoid(0) = 0.5.
# So, 2 * 0.5 = 1.

def custom_activation_sls(x):
    return np.where(x >= 0, x, 2 * np.log(1 + np.exp(x)))

# Generate input range
x_custom = np.linspace(-10, 10, 400)
y_custom = custom_activation_sls(x_custom)

# Plotting the custom activation function
plt.figure(figsize=(8, 6))
plt.plot(x_custom, y_custom, label='Smooth Linear Softplus (SLS)')
plt.plot(x_custom, relu(x_custom), linestyle='--', alpha=0.7, label='ReLU (for comparison)')

plt.title('Custom Activation Function: Smooth Linear Softplus (SLS)')
plt.xlabel('Input (x)')
plt.ylabel('Output f(x)')
plt.grid(True)
plt.legend()
plt.axvline(0, color='black', linestyle='--', linewidth=0.8)
plt.axhline(0, color='black', linestyle='--', linewidth=0.8)
plt.ylim(-2, 10) # Adjust y-limit for better visualization
plt.show()

print("Custom Activation Function: Smooth Linear Softplus (SLS) has been defined and plotted.")
```

```python
print("\n--- Task 8: Advantages and Comparison with ReLU ---")

print("Advantages of Smooth Linear Softplus (SLS) compared to ReLU:")
print("1. Smoothness: SLS is a continuously differentiable function across its entire domain, unlike ReLU which has a non-differentiable point at x=0. Smoothness is often beneficial for gradient-based optimization.")
print("2. No 'Dead ReLU' Problem: For negative inputs, SLS provides a non-zero, positive gradient (2 * sigmoid(x)). This ensures that neurons with negative inputs can still learn and contribute to the gradient, mitigating the 'dead ReLU' problem where neurons can get stuck for negative inputs.")

print("Comparison with ReLU:")
print("- Positive Region (x >= 0): Both SLS and ReLU behave identically, returning f(x) = x, which helps avoid vanishing gradients for positive inputs.")
print("- Negative Region (x < 0): SLS provides a smooth, exponentially saturating output with a positive gradient, while ReLU outputs 0 with a zero gradient. This is the key difference in handling negative activations.")
print("- Gradient at x=0: SLS is differentiable at x=0 with a gradient of 1, providing a smooth transition. ReLU has an abrupt change in gradient at x=0 (from 0 to 1). ")
```
